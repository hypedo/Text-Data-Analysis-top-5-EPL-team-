````python
# =====================================================
# 0. 라이브러리 / 파일 설정
# =====================================================

import re
import json
import time
import pandas as pd
from tqdm import tqdm

# 이미 OpenAI client를 만들었다면 생략
from openai import OpenAI
client = OpenAI()

INPUT_FILE = "liverpool_reanalyze_target.xlsx"

OUTPUT_XLSX = "liverpool_reanalyze_target_labeled.xlsx"
OUTPUT_CSV = "liverpool_reanalyze_target_labeled.csv"


# =====================================================
# 1. GPT 응답 JSON 파싱 함수
# =====================================================

def parse_json_array(text):
    text = text.strip()

    # 코드블록 제거
    text = re.sub(r"^```json|^```|```$", "", text, flags=re.MULTILINE).strip()

    # JSON 배열 부분만 추출
    start = text.find("[")
    end = text.rfind("]")

    if start != -1 and end != -1:
        text = text[start:end+1]

    return json.loads(text)


# =====================================================
# 2. 데이터 불러오기
# =====================================================

df = pd.read_excel(INPUT_FILE)

df = df.reset_index(drop=True).copy()
df["analysis_id"] = df.index

df["text_clean"] = df["text_clean"].fillna("").astype(str)
df = df[df["text_clean"].str.len() >= 2].reset_index(drop=True)

print("분석 대상 데이터 수:", len(df))
print("\n[source 분포]")
print(df["source"].value_counts(dropna=False))

print("\n[content_type 분포]")
print(df["content_type"].value_counts(dropna=False))

print("\n[gpt_sentiment 기존 분포]")
print(df["gpt_sentiment"].value_counts(dropna=False))
````

````python
# =====================================================
# 3. match_context 생성용 라운드별 경기 맥락
# =====================================================
# 여기에 리버풀 29R~38R 경기 결과/맥락을 직접 입력

match_context_map = {
    "29R": "리버풀이 울버햄튼에게 경기 막판에 2:1로 패배",
    "30R": "리버풀이 토트넘에게 경기 막판 골을 허용해서 1:1 무승부",
    "31R": "리버풀이 브라이튼에게 2:1로 패배",
    "32R": "리버풀이 풀럼을 상대로 2:0으로 승리",
    "33R": "리버풀이 에버턴을 상대로 막판에 극장골을 넣고 2:0으로 승리",
    "34R": "리버풀이 크리스탈 팰리스를 상대로 3:1로 승리",
    "35R": "리버풀이 맨체스터 유나이티드에게 3:2로 패배",
    "36R": "리버풀이 첼시와 1:1 무승부",
    "37R": "리버풀이 아스톤 빌라에게 4:2로 패배",
    "38R": "리버풀이 살라와 로버트슨의 고별전에서 브렌트포드와 1:1로 무승부",
}
````

````python
# =====================================================
# 4. match_context 생성 함수
# =====================================================

def make_match_context(row):
    source = row.get("source")
    round_info = row.get("round")
    title = row.get("title")
    period = row.get("period")
    content_type = row.get("content_type")
    content_published_at = row.get("content_published_at")
    comment_published_at = row.get("comment_published_at")

    # 유튜브: 라운드별 경기 맥락 사용
    if source == "youtube":
        base_context = match_context_map.get(
            round_info,
            "리버풀 관련 유튜브 영상 댓글. 영상 제목과 댓글 내용을 바탕으로 판단."
        )

        return f"""
{base_context}

[데이터 정보]
데이터 출처: YouTube 댓글
영상 제목: {title}
라운드: {round_info}
영상 업로드일: {content_published_at}
댓글 작성일: {comment_published_at}
38R 기준 구분: {period}

이 댓글은 리버풀 관련 유튜브 영상에 달린 댓글이다.
영상 제목과 라운드 정보를 댓글 해석에 반영한다.
"""

    # 카페: 경기 하이라이트가 아니라 팬 커뮤니티 의견
    if source == "cafe":
        if period == "before_38R":
            return f"""
[데이터 정보]
데이터 출처: 리버풀 팬 카페
게시글 제목: {title}
데이터 유형: {content_type}
게시글 작성일: {content_published_at}
댓글 작성일: {comment_published_at}
38R 기준 구분: before_38R

이 텍스트는 38R 이전 리버풀 팬 카페 게시글 또는 댓글이다.
시즌 종료 직전 팬들의 기대, 우려, 경기 전 전망, 선수단 평가, 감독/전술 평가가 포함될 수 있다.
영상 장면에 대한 즉각 반응이 아니라 팬 커뮤니티의 의견으로 해석한다.
"""
        elif period == "after_38R":
            return f"""
[데이터 정보]
데이터 출처: 리버풀 팬 카페
게시글 제목: {title}
데이터 유형: {content_type}
게시글 작성일: {content_published_at}
댓글 작성일: {comment_published_at}
38R 기준 구분: after_38R

이 텍스트는 38R 이후 리버풀 팬 카페 게시글 또는 댓글이다.
시즌 종료 직후 팬들의 시즌 총평, 선수 평가, 감독 평가, 이적시장 의견, 구단 운영 평가가 포함될 수 있다.
영상 장면에 대한 즉각 반응이 아니라 팬 커뮤니티의 의견으로 해석한다.
"""
        else:
            return f"""
[데이터 정보]
데이터 출처: 리버풀 팬 카페
게시글 제목: {title}
데이터 유형: {content_type}

이 텍스트는 리버풀 팬 카페 게시글 또는 댓글이다.
팬 커뮤니티의 의견으로 해석한다.
"""

    return "리버풀 관련 텍스트 데이터."


df["match_context"] = df.apply(make_match_context, axis=1)

# 같은 맥락끼리 묶기 위한 key
df["context_key"] = (
    df["source"].astype(str) + " | " +
    df["round"].astype(str) + " | " +
    df["period"].astype(str) + " | " +
    df["title"].astype(str)
)

print("match_context 결측 수:", df["match_context"].isna().sum())

display(
    df[[
        "source",
        "round",
        "period",
        "title",
        "content_type",
        "text_clean",
        "match_context"
    ]].head()
)
````

````python
# =====================================================
# 5. 1차 감정분석 함수
# =====================================================

def analyze_batch(comments, match_context):
    comments_text = "\n".join([
        f"{i}. {comment}"
        for i, comment in enumerate(comments)
    ])

    prompt = f"""
너는 한국어 축구 팬덤 댓글을 분석하는 target-aware 감정분석 분류기다.
분석 대상은 EPL 리버풀 관련 유튜브, 카페 댓글이다.

[리버풀 선수단 정보]
현재 감독: 아르네 슬롯
이전 감독: 위르겐 클롭

선수:
알리송 베케르, 조 고메즈(고메스), 엔도 와타루, 버질 반다이크, 코나테, 케르케즈, 비르츠,
소보슬러이(소보슬라이), 이삭(이사크), 맥 알리스터, 모하메드 살라, 브래들리,
키에사, 레오니, 커티스 존스, 각포(학포), 에키티케, 마마르다슈빌리(마마슈), 로버트슨
우드먼, 프림퐁, 흐라벤베르흐(그라벤베르흐), 트레이 뇨니, 리오 은구모하,

[선수단 반영 규칙]
- 위 선수나 감독을 칭찬하면 리버풀 관련 positive로 분류한다.
- 위 선수나 감독을 비판, 조롱, 실망하면 리버풀 관련 negative로 분류한다.
- 위 선수 이름이 직접 나오면 target은 보통 liverpool_player로 둔다.
- 위르겐 클롭, 아르네 슬롯 관련 댓글이면 target은 liverpool_manager로 둔다.
- 리버풀 이외의 다른 대상을 언급한 댓글은 리버풀 감정으로 보지 않고 neutral로 분류한다.

[영상/게시글/경기 맥락]
{match_context}

이 맥락은 댓글 해석에 반드시 반영한다.
리버풀이 승리한 경기에서는 "골 미쳤다", "마무리 좋다", "슈팅 지렸다", "수비 좋다"처럼 대상이 불명확한 경기력 칭찬은 특별히 상대팀 언급이 없으면 리버풀 positive로 우선 판단한다.
리버풀이 패배한 경기에서는 아쉬움, 비판, 조롱, 실망 표현이 리버풀을 향하면 negative로 판단한다.

[분류 기준]
- positive: 리버풀 팀/선수/감독/전술/경기력/결과에 대한 칭찬, 감탄, 기대, 축하
- negative: 리버풀 팀/선수/감독/전술/경기력/결과에 대한 비판, 실망, 조롱, 비꼼, 분노
- neutral: 리버풀과 직접 관련 없거나 단순 질문/정보/감정 판단 불가

[중요 규칙]
- neutral은 정말 감정 방향을 판단할 수 없을 때만 사용한다.
- 리버풀 관련 맥락이 있고 긍정/부정 뉘앙스가 조금이라도 있으면 positive 또는 negative로 판단한다.
- "미쳤다", "지린다", "개쩐다", "폼 미쳤다", "찢었다", "월클", "GOAT"는 칭찬 맥락이면 positive.
- "개못한다", "답답하다", "망했다", "처참하다", "노답", "왜 저러냐"는 비판 맥락이면 negative.
- 상대팀/상대선수 칭찬은 리버풀과 무관하면 neutral.
- 단, 상대팀 칭찬이 리버풀의 패배/부진/약점을 강조하면 negative.
- 단순 감탄사라도 댓글 맥락상 리버풀 경기 장면에 대한 반응으로 해석 가능하면 neutral보다 positive/negative를 우선한다.

[target 기준]
- liverpool_team
- liverpool_player
- liverpool_manager
- opponent
- referee_or_league
- unclear

[스코어/라운드 맥락 활용]
라운드 정보와 스코어는 댓글 해석에 반영한다.
match_context에서 상대팀이 득점하지 못한 경기라고 주어진 경우,
"골 미쳤다", "마무리 좋다", "슈팅 지렸다", "득점 장면 좋다"처럼 득점/공격 장면을 칭찬하는 댓글은
특별히 상대팀이나 상대 선수가 명시되지 않는 한 리버풀 positive로 판단한다.

리버풀이 패배한 경기에서는 경기력 비판, 실망, 조롱, 비꼼 표현을 negative로 판단한다.
단, 패배 경기라도 리버풀 선수 개인의 좋은 플레이를 칭찬하면 positive로 판단한다.

[출력 규칙]
반드시 JSON 배열만 출력하라.
설명 문장, 코드블록, 마크다운은 쓰지 마라.
각 결과에는 입력 번호 id를 포함하라.

[
  {{
    "id": 0,
    "sentiment": "positive 또는 negative 또는 neutral",
    "confidence": 0.0부터 1.0 사이 숫자,
    "target": "liverpool_team 또는 liverpool_player 또는 liverpool_manager 또는 opponent 또는 referee_or_league 또는 unclear",
    "reason": "짧은 한국어 이유"
  }}
]

[분석할 댓글]
{comments_text}
"""

    try:
        response = client.chat.completions.create(
            model="gpt-4.1-mini",
            messages=[
                {"role": "system", "content": "너는 한국어 축구 댓글 감정분석 전문가다."},
                {"role": "user", "content": prompt}
            ],
            temperature=0
        )

        result_text = response.choices[0].message.content.strip()
        return parse_json_array(result_text)

    except Exception as e:
        return [
            {
                "id": i,
                "sentiment": "error",
                "confidence": 0.0,
                "target": "error",
                "reason": str(e)
            }
            for i in range(len(comments))
        ]
````

````python
# =====================================================
# 6. 2차 neutral 재분석 함수
# =====================================================

def analyze_neutral_batch(comments, match_context):
    comments_text = "\n".join([
        f"{i}. {comment}"
        for i, comment in enumerate(comments)
    ])

    prompt = f"""
너는 리버풀 관련 유튜브, 카페 댓글의 2차 감정분석 분류기다.
아래 댓글들은 1차 분석에서 neutral로 분류되었지만, neutral 판정을 최소화하기 위해 재검토한다.

[맥락]
{match_context}

[가장 중요한 원칙]
neutral은 예외값이다.
감정 방향이 정말 전혀 없을 때만 neutral로 둔다.

칭찬, 감탄, 만족, 축하, 기대가 조금이라도 있으면 positive.
비판, 실망, 조롱, 분노, 답답함, 비꼼이 조금이라도 있으면 negative.

댓글이 짧거나 대상이 생략되어도,
경기 맥락상 리버풀에 대한 반응으로 자연스럽게 해석되면
positive 또는 negative로 분류한다.

[neutral로 유지하는 경우]
- 단순 질문
- 경기 시간, 일정, 정보 확인
- 영상 편집, 중계, 해설, 업로드 관련 이야기
- 리버풀 / 리버풀 선수 / 감독 / 경기 결과와 감정적으로 연결되지 않는 말
- 감정 단서가 전혀 없는 단순 사실 전달

[neutral 금지 예시]
"미쳤다", "지렸다", "좋다", "깔끔하다", "나이스", "레전드", "GOAT" → positive
"답답하다", "망했다", "왜 저러냐", "최악", "또 이러네", "아쉽다" → negative

"역시 리버풀"은 칭찬이면 positive, 조롱/비꼼이면 negative.
상대팀 칭찬이라도 리버풀의 패배, 약점, 부진을 강조하면 negative.

[리버풀 팬덤 멸칭 및 조롱 표현 규칙]

다음 표현은 리버풀을 조롱하거나 비하하는 표현으로 간주한다.

아래 단어가 포함되면 문맥과 관계없이 sentiment를 negative로 분류한다.

- 리중딱
- 살인풀
- 전범풀
- 뎀바바
- 소라드

또한 "훔" 또는 "훔바"로 시작하는 단어가 포함되면 sentiment를 negative로 분류한다.

예시:
- 리중딱
- 살인풀 또 시작이네
- 전범풀 수준
- 뎀바바는 영원한 레전드
- 소라드 또 날렸네
- 훔바훔바
- 훔○○

위 표현들은 리버풀에 대한 조롱, 비하, 멸칭으로 간주하며 반드시 negative로 분류한다.

[소보슬라이(소보슬러이) 조롱 표현 규칙]

다음 조건에 해당하면 sentiment를 negative로 분류한다.

- 소보슬라이 또는 소보슬러이를 제라드와 비교하며 넘어짐, 미끄러짐, 실수 장면을 조롱하는 경우
- 소보슬라이 또는 소보슬러이를 제라드의 유명한 미끄러짐 장면과 연결하여 비하하는 경우
- 소보슬라이 또는 소보슬러이의 실수, 넘어짐, 미끄러짐을 웃음거리로 만드는 경우
- "제라드 후계자", "제2의 제라드", "제라드 빙의", "제라드 DNA", "제라드 환생" 등의 표현이 조롱 의도로 사용되는 경우
- "미끄덩", "또 넘어지네", "슬라이딩", "넘어져서 골 먹힘" 등과 함께 소보슬라이(소보슬러이)가 언급되는 경우

예시:sh

- 소보슬라이 제라드 후계자네 ㅋㅋ
- 소보슬러이도 결국 제라드 과학
- 제2의 제라드 탄생
- 또 미끄러졌네
- 제라드 DNA 어디 안 간다
- 소보슬라이 넘어지는 거 개웃기네
- 소보슬러이 미끄덩 ㅋㅋ
- 제라드 환생했네

위와 같은 표현은 리버풀 선수에 대한 조롱 및 비하로 간주하며 반드시 negative로 분류한다.

[조롱성 웃음 표현 규칙]

댓글에 "ㅋㅋ", "ㅋㅋㅋ", "개웃기네", "존나 웃기네", "웃음벨", "개꿀잼", "레전드", "과학" 등이 포함되어 있더라도

리버풀 또는 리버풀 선수의 실수, 패배, 미끄러짐, 부진, 실책을 비웃는 맥락이면 반드시 negative로 분류한다.

예시:

- 또 넘어졌네 ㅋㅋㅋ
- 소보슬라이 과학이네
- 살라 폼 다 죽었네 ㅋㅋ
- 리버풀 또 졌네 개웃기네

[판단 우선순위]
1. 비판 / 조롱 / 실망 단서가 있으면 negative
2. 칭찬 / 감탄 / 만족 단서가 있으면 positive
3. 둘 다 전혀 없을 때만 neutral

[출력 규칙]
반드시 JSON 배열만 출력한다.
설명 문장, 코드블록, 마크다운은 출력하지 않는다.

[
  {{
    "id": 0,
    "sentiment": "positive 또는 negative 또는 neutral",
    "confidence": 0.0,
    "target": "liverpool_team 또는 liverpool_player 또는 liverpool_manager 또는 opponent 또는 referee_or_league 또는 unclear",
    "reason": "짧은 한국어 이유"
  }}
]

[재분석할 댓글]
{comments_text}
"""

    try:
        response = client.chat.completions.create(
            model="gpt-4.1-mini",
            messages=[
                {
                    "role": "system",
                    "content": "너는 한국어 축구 댓글 감정분석 전문가다. neutral 판정은 매우 보수적으로 사용한다."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=0
        )

        result_text = response.choices[0].message.content.strip()
        return parse_json_array(result_text)

    except Exception as e:
        return [
            {
                "id": i,
                "sentiment": "neutral",
                "confidence": 0.0,
                "target": "unclear",
                "reason": str(e)
            }
            for i in range(len(comments))
        ]
````

````python
# =====================================================
# 7. 1차 감정분석 실행
# =====================================================

def run_first_pass(df, batch_size=20):
    final_df = df.copy()

    final_df["new_gpt_sentiment"] = None
    final_df["new_gpt_confidence"] = None
    final_df["new_gpt_target"] = None
    final_df["new_gpt_reason"] = None

    for context_key, group_df in tqdm(final_df.groupby("context_key"), desc="1차 감정분석"):
        match_context = group_df["match_context"].iloc[0]

        group_df = group_df.reset_index()

        for start in range(0, len(group_df), batch_size):
            batch_df = group_df.iloc[start:start+batch_size]
            comments = batch_df["text_clean"].tolist()

            batch_results = analyze_batch(comments, match_context)

            result_map = {}

            for r in batch_results:
                try:
                    result_map[int(r.get("id"))] = r
                except:
                    pass

            for local_id, row in enumerate(batch_df.itertuples()):
                r = result_map.get(local_id, {})

                idx = row.index

                sentiment = str(r.get("sentiment", "error")).lower().strip()

                if sentiment not in ["positive", "negative", "neutral", "error"]:
                    sentiment = "error"

                final_df.at[idx, "new_gpt_sentiment"] = sentiment
                final_df.at[idx, "new_gpt_confidence"] = r.get("confidence", 0.0)
                final_df.at[idx, "new_gpt_target"] = r.get("target", "unclear")
                final_df.at[idx, "new_gpt_reason"] = r.get("reason", "")

            time.sleep(0.3)

    return final_df


labeled_df = run_first_pass(df, batch_size=20)

print("1차 감정분석 결과")
print(labeled_df["new_gpt_sentiment"].value_counts(dropna=False))
````

````python
# =====================================================
# 8. 2차 neutral 재분석
# =====================================================

def run_second_pass_for_neutral(labeled_df, batch_size=20):
    final_df = labeled_df.copy()

    final_df["gpt_sentiment_1st"] = final_df["new_gpt_sentiment"]
    final_df["gpt_reason_1st"] = final_df["new_gpt_reason"]
    final_df["neutral_second_pass"] = False

    neutral_df = final_df[
        final_df["new_gpt_sentiment"].astype(str).str.lower().eq("neutral")
    ].copy()

    print("2차 재분석 대상 neutral 수:", len(neutral_df))

    for context_key, group_df in tqdm(neutral_df.groupby("context_key"), desc="2차 neutral 재분석"):
        match_context = group_df["match_context"].iloc[0]

        group_df = group_df.reset_index()

        for start in range(0, len(group_df), batch_size):
            batch_df = group_df.iloc[start:start+batch_size]
            comments = batch_df["text_clean"].tolist()

            batch_results = analyze_neutral_batch(comments, match_context)

            result_map = {}

            for r in batch_results:
                try:
                    result_map[int(r.get("id"))] = r
                except:
                    pass

            for local_id, row in enumerate(batch_df.itertuples()):
                r = result_map.get(local_id, {})

                idx = row.index

                sentiment = str(r.get("sentiment", "neutral")).lower().strip()

                if sentiment not in ["positive", "negative", "neutral"]:
                    sentiment = "neutral"

                final_df.at[idx, "new_gpt_sentiment"] = sentiment
                final_df.at[idx, "new_gpt_confidence"] = r.get("confidence", 0.0)
                final_df.at[idx, "new_gpt_target"] = r.get("target", "unclear")
                final_df.at[idx, "new_gpt_reason"] = "[2차 neutral 재분석] " + str(r.get("reason", ""))
                final_df.at[idx, "neutral_second_pass"] = True

            time.sleep(0.3)

    return final_df


labeled_df = run_second_pass_for_neutral(labeled_df, batch_size=20)

print("2차 재분석 후 결과")
print(labeled_df["new_gpt_sentiment"].value_counts(dropna=False))
````
